Q5. Melt and Pivot — Medium

```python
student_record = pd.DataFrame({'Student_ID':[1,2,3,4,5,6], 
'Name':['Ali','Bahadr','Hassan','Hamza','Sana','Sara'],
'Class':['A','A','A','B','B','B'],
'Math':[10,20,30,40,50,60],
'English':[10,20,30,40,50,60],
'Science':[10,20,30,40,50,60]})

student_record

```

```python
student_record1 = pd.melt(student_record,id_vars=['Student_ID', 'Name', 'Class'], 
value_vars=['Math', 'English', 'Science'],
var_name='Subject', value_name='Marks')

student_record1
```

```python
student_record1.pivot(index=['Student_ID', 'Name', 'Class'], columns='Subject', values='Marks')
```

Merge, Join and Concatenate — Hard

```python
customer_record = pd.DataFrame({'Customer_ID':[1,2,3,4,5,6],
'Customer_Name':['Ali','Bahadr','Hassan','Hamza','Sana','Sara'],
'City':['Lahore','Karachi','Islamabad','Rawalpindi','Faisalabad','Bahawalpur'],
})

customer_record
```

```python
sales_record = pd.DataFrame({'Sales_ID':[1,2,3,4,5,6],
'Customer_ID':[1,2,3,4,5,6],
'Product':[1,2,3,4,5,6],
'Amount':[10,20,30,40,50,60]})

sales_record
```

```python
new_customer_record = pd.DataFrame({'Customer_ID':[7,8,9,10,11,12],
'Customer_Name':['Ali','Bahadr','Hassan','Hamza','Sana','Sara'],
'City':['Lahore','Karachi','Islamabad','Rawalpindi','Faisalabad','Bahawalpur']})

new_customer_record
```

```python
# a) Use merge() to combine the Customer and Sales DataFrames using "Customer_ID" .
customer_record.merge(sales_record, on='Customer_ID')
```

```python
customer_record.set_index('Customer_ID')
```

```python
sales_record.set_index('Customer_ID')
```

```python
customer_record.join(sales_record.set_index('Customer_ID'), on='Customer_ID')
```

```python
combined = pd.concat([customer_record, new_customer_record], ignore_index=True)
combined
```
