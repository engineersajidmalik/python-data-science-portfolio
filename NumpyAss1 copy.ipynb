Question 23 Given: arr = np.arange(1, 37).reshape(6, 6) Use slicing to extract: (a) top-left 3 × 3 section, (b) top-right 3 × 3 section, (c) bottom-left 3 × 3 section, (d) bottom-right 3 × 3 section

```python
arr = np.arange(1,37).reshape(6,6)
print(arr)
# top-left 3 × 3 section
print(arr[0:3:,0:3:])
#  top-left 3 × 3 section
print(arr[0:3:,3::])
#  bottom-left 3 × 3 section
print(arr[3::,0:3])
#  bottom-right 3 × 3 section
print(arr[3::,3::])
```

Given: arr = np.arange(1, 37).reshape(6, 6) Use slicing to select: (a) every second row, (b) every second column, (c) every second row and every second column. 

```python
arr = np.arange(1,37).reshape(6,6)
print(arr)

# Every 2nd row
print(arr[0::2,])

# Every 2nd column
print(arr[:,0::2])

# Every 2nd row and every 2nd column
print(arr[0::2,0::2])
```

```python
arr = np.arange(1, 51).reshape(10, 5)
print(arr)

# Last 4 rows first 2 col
print("\n")
print(arr[-4::,0:2:])

# Every 3rd row and all col
print("\n")
print(arr[0::3,0::])

# The 3×3 submatrix in the exact center of the array
print("\n")
print(arr[4:7:,1:4:])

# A single "checkerboard-style" slice: every second row and every 
# second column, starting from row 1 and column 1 (not row 0)
print("\n")
print(arr[1::2,1::2])
```
