scatter tells the realtions between 2 variables 

```python
plt.scatter(tips['total_bill'], tips['tip'])
```

```python
fig = plt.figure(figsize=(10,6))
plt.scatter(tips['total_bill'], tips['tip'], c = 'violet', marker='*')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.title('Scatter plot of Total Bill vs Tip')
plt.legend()
plt.show()
```

```python
sp = sns.scatterplot(tips,x="total_bill",y="tip", hue="sex",size="tip")
sp
```
