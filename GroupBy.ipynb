```python
sales_data = pd.read_csv('Datasets//sales_data.csv')
sales_data.sort_index(inplace=True)
sales_data.head()
```

```python
pratice = pd.read_csv('Datasets/sales_data.csv')
pratice.sort_index(inplace=True)
pratice.head()
```

```python
# Sales of each City
pratice.groupby('City', as_index=True)['Sales'].sum()
```

```python
# Average quantity of each product
pratice.groupby('Product')['Quantity'].mean()
```

```python
pratice.groupby('Category').count()

pratice.groupby('Category').size()
```

```python
pratice.groupby('Salesperson')['Sales'].sum().sort_values(ascending=False)
```

```python
pratice.groupby(['City', 'Product'])['Sales'].sum()
```

```python
pratice.groupby('Salesperson')['Sales'].agg(['max', 'min'])
```
