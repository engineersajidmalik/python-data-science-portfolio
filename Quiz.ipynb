Create a labeled DataFrame 4x3 from a dictionary, with indexes named Row1, Row2, … and columns named Col1, Col2, …. Then add one column and drop one row from it.



```python
labelled_data = pd.DataFrame.from_dict({'Row1':[1, 2, 3], 'Row2':[4, 5, 6], 'Row3':[7, 8, 9],'Row4':[10, 11, 12]}, 
 orient='index', columns=['Col1', 'Col2', 'Col3'])

labelled_data
```

```python
# adding one col
labelled_data['Col4'] = [13, 14, 15, 16]
labelled_data
```

```python
# Drop one row
dropped = labelled_data.drop('Row2', axis=0, inplace=False)
dropped

```
