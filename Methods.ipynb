```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# Long DataFormat and Wide DataFormat

```python
wide_data = pd.DataFrame({'Std_ID': [1, 2, 3, 4, 5],'Marks': [10, 20, 30, 40, 50],
'Fee': [1000, 2000, 3000, 4000, 5000], 'Subject': ['Math', 'Physics', 'Chemistry', 'Biology', 'English'],
'Age': [15, 16, 17, 18, 19]})

wide_data
```

```python
# Now we will convert the wide data into long data using melt()
long_data = wide_data.melt(id_vars=['Std_ID', 'Subject'], var_name='Variable', value_name='Value')
long_data
```

```python
# Now we will convert the long data into wide data using pivot()
long_data.pivot(index=['Std_ID', 'Subject'], columns='Variable', values='Value').reset_index()
```

# Merge Funtion

```python
data1 = pd.DataFrame({'Std_ID': [1, 2, 3, 4, 5],'Marks': [10, 20, 30, 40, 50],
'Fee': [1000, 2000, 3000, 4000, 5000], 'Subject': ['Math', 'Physics', 'Chemistry', 'Biology', 'English'],
'Age': [15, 16, 17, 18, 19]})
data1
```

```python
data2 = pd.DataFrame({'Std_ID': [1, 2, 3, 4, 5],'Marks': [10, 20, 30, 40, 50],
'Charges': [1000, 2000, 3000, 4000, 5000], 'Course': ['Math', 'Physics', 'Chemistry', 'Biology', 'English'],
'Old': [15, 16, 17, 18, 19]})
data2
```

```python
data3 = pd.concat([data1, data2], axis=0, ignore_index=True)
data3
```

```python
# Now will apply the merge() method to combine data1 and data2 based on the common column 'Std_ID'
merged_data = pd.merge(data1, data2, on='Std_ID', how='right')
merged_data
```
