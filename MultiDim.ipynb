How can we set explicit data type in NumPy

```python
import numpy as np
np
```

```python
arr = np.array([1,2,3,4,5], dtype=np.int8)
print(arr.dtype)
print(arr.itemsize)
```

Slicing
arr[start:stop:step]

```python
vector = np.arange(30)
vector
```

```python
vector[::]
```

```python
vector[::-1]
```

```python
vector[5::2]
```

Multidimensional Slicing
tensor[sheet:row:col]

```python
tensor = np.arange(300).reshape(20,3,5)
tensor
```

```python
matrix = np.arange(30).reshape(5,6)
matrix
```

```python
matrix[1:4,2:4]
```

```python
matrix[1:4,]
```
