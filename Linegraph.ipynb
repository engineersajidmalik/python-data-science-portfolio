```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
flights = sns.load_dataset('flights')
flights.head()
```

```python
sns.lineplot(x='year', y='passengers',data=flights, hue='month')


```

```python
sc = sns.scatterplot(flights, x='year', y='passengers', hue='month')
plt.figure(figsize=(10,6))
plt.xlabel('Year')
plt.ylabel('Passengers')
plt.title('Scatter plot of Year vs Passengers')
plt.legend(loc='center right', bbox_to_anchor=(1.0, 0.5))
plt.show()
```

```python
data = sns.load_dataset('flights')
data.head()
```

```python
# Now we will use line plot to show the graph is connected with license
plt.figure(figsize=(6,3))
sns.lineplot(
x='year', 
y='passengers',
data=data, 
hue='month',
style='month',
dashes=False,
markers=True
)
plt.xlabel('year')
plt.ylabel('passengers')
plt.title('Graph of years and Passangers ')
plt.legend(loc='center left', bbox_to_anchor=(1.0, 0.5))
plt.show()
```

```python
tips = sns.load_dataset('tips')
tips.head()
```

```python
plt.figure(figsize=(6,3))
sns.scatterplot(data= tips, 
x='total_bill', 
y='tip', hue='time',
size='size',
style='smoker'


)
plt.xlabel('total_bill')
plt.ylabel('tip')
plt.title('Scatter plot of Total Bill vs Tip')
plt.legend(loc='center left', bbox_to_anchor=(1.0, 0.5))
plt.show()
```

```python
f = sns.load_dataset('flights')
f.head()
```

```python
plt.figure(figsize=(6,3))
may_flights = f.query('month == "May"')
sns.lineplot(data=may_flights, x='year', y='passengers')
```

Now we will use pivot to change data foramat

```python
flight_wide = f.pivot(index='year', columns='month', values='passengers')
flight_wide.head()
```

```python
plt.figure(figsize=(6,3))
sns.lineplot(data=flight_wide['Apr'])
```

```python
plt.figure(figsize=(6,3))
sns.lineplot(data=flight_wide[['Apr','May','Sep']])
plt.legend(loc='center left', bbox_to_anchor=(1.0, 0.5))
```
