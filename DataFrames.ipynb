```python
df_movies1 = pd.read_csv('Datasets/imdb_top_1000.csv ')
df_movies1.head()
# By default head will take only first 5
```

```python
# Now we will apply loc and iloc for filtering data on rows and columns 
# loc will get lable index 
df_movies1.loc[0:5:,'Poster_Link':'Certificate']
```

```python
# Now we will apply iloc that will get numeric index
df_movies1.iloc[0:3:,0:3]
```
