```python
import numpy as np
import pandas as pd
```

Alignment 


```python
align = pd.Series([90,29,39,47,50], 
index=['English','Math', 'Physics', 'Chemistry', 'Biology'], dtype='int8')

align
```

```python
align1 = pd.Series([90,29,39,47,50], 
index=['English','Physics', 'Chemistry', 'Isl','Biology'], dtype='int8')

align1
```

Golden Rule in Alignment:
If label is match = operation is performed
if lable is present in one obj and missing in other obj = NaN will be returned

```python
align2 = align + align1
align2
```

```python
align3 = align.add(align1, fill_value=0)
align3
```

```python
align3.isnull()
```

```python
align3.notnull()
```
