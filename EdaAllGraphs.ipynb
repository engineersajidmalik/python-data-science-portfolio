```python
titanic = sns.load_dataset('titanic')
titanic.head()
```

```python
fig, ax = plt.subplots(figsize=(6,3))
ax.hist(titanic['age'], bins=10, color='red', edgecolor='black')
ax.set_xlabel('Age')
ax.set_ylabel('Count')
ax.set_title('Histogram of Passangers')
plt.show()
```

```python
# Now we will use density of histogram
fig, ax = plt.subplots(figsize=(6,3))
ax.hist(titanic['age'], density=True, bins=10, color='red', edgecolor='black')
ax.set_xlabel('Age')
ax.set_ylabel('Count')
ax.set_title('Histogram of Passangers')
plt.show()
```

```python
age = titanic['age'].dropna()
plt.figure(figsize=(6,3))
plt.boxplot(age)
plt.xlabel('Age')
plt.ylabel('Count')
plt.title('Boxplot of Passangers')
plt.show()
```

```python
penguins = sns.load_dataset('penguins')
penguins.head()
```

```python
plt.figure(figsize=(6,3))
plt.hist(penguins['body_mass_g'],bins=15, color='skyblue', edgecolor='black')
plt.title('Histogram of Body Mass')
plt.xlabel('Body Mass')
plt.ylabel('Count')
plt.show()
```

```python
# Boxplot of bodymass
plt.figure(figsize=(6,3))
sns.boxplot(penguins['body_mass_g'])
sns.set_style("whitegrid")
plt.xlabel('Body Mass')
plt.ylabel('Count')
plt.title('Boxplot of Body Mass')
plt.show()
```

```python
plt.figure(figsize=(6,3))
sns.kdeplot(data=penguins['body_mass_g'], shade=True, cbar=True, fill=True)
plt.xlabel('Body Mass')
plt.ylabel('Count')
plt.title('Density Plot of Body Mass')
plt.show()
```

```python
plt.figure(figsize=(6,3))
sns.kdeplot(data=penguins, x='body_mass_g', hue='species', shade=True, cbar=True, fill=True)
plt.xlabel('Body Mass')
plt.ylabel('Count')
plt.title('Density Plot of Body Mass')
plt.show()
```

```python
sns.boxplot(data=penguins, x='body_mass_g', hue='species')
```

```python
species = penguins['species'].value_counts()
species
```

```python
plt.figure(figsize=(6,3))
plt.bar(species.index, species.values)
plt.xlabel('Species')   
plt.ylabel('Count')
plt.title('Barplot of Species')
plt.show()
```

```python
plt.figure(figsize=(6,3))
plt.barh(species.index, species.values)
plt.xlabel('Species')   
plt.ylabel('Count')
plt.title('Barplot of Species')
plt.show()
```

```python
plt.figure(figsize=(6,3))
plt.pie(species.values, labels=species.index, autopct='%1.1f%%')
```

```python
# explode is used for space in each row
plt.figure(figsize=(6,3))
plt.pie(species.values, labels=species.index,explode=[0.1,0.1,0], autopct='%1.1f%%')
```

```python
plt.figure(figsize=(6,3))
sns.violinplot(data=penguins, x='body_mass_g', hue='species')
```

# Relational Plot

```python
plt.figure(figsize=(6,3))
sns.relplot(data=penguins, x='flipper_length_mm', y='body_mass_g', hue='species')
```

```python
sns.relplot(data=penguins, x='flipper_length_mm', y='body_mass_g', hue='species', kind='scatter', col='sex')
```

```python
# Cat plot
plt.figure(figsize=(6,3))
sns.catplot(data=tips, y='total_bill', row='sex', kind='swarm', col='smoker')
```

```python
g = sns.FacetGrid(data=tips, col='sex', row='smoker')
g.map(plt.scatter, 'total_bill', 'tip')
```

```python
# pair plot
sns.pairplot(data=penguins, hue='species')
```

```python
sns.pairplot(data=penguins)
```

```python
g = sns.PairGrid(data=penguins, hue='species')
g.map(plt.scatter)
```

```python
sns.pairplot(data=penguins, hue='species', diag_kind='kde')
```

```python
g = sns.pairplot(data=penguins, hue='species')
g.map(plt.scatter)
g.map_diag(sns.kdeplot)
```

```python
g = sns.pairplot(data=penguins, hue='species')
g.map_offdiag(plt.scatter)
g.map_diag(sns.kdeplot)
```

```python
sns.rugplot(data=penguins, x='body_mass_g')
```
