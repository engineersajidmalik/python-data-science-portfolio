```python
class MyMetaClass(type):
    def __new__(cls, name, bases, namespace, /, **kwds):
        print('MyMetaClass.__new__', name, bases, namespace, kwds)
        
        return super().__new__(cls,name, bases, namespace, **kwds)
    
    def __init__(self, name, bases, namespace, /, **kwds):
        print('MyMetaClass.__init__', name, bases, namespace, kwds)
        super().__init__(name, bases, namespace, **kwds)
        
    def __call__(self, *args, **kwargs):
        print('MyMetaClass.__call__', args, kwargs)
        return super().__call__(*args, **kwargs)
    
```

```python
class Student(metaclass=MyMetaClass):
    def __new__(cls):
        pass
    def __init__(self, name):
        self.name = name
        print('Student.__init__', name)
       
   
```

```python
print(type(Student))
```
